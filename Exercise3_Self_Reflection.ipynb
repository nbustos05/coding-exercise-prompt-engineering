# Exercise 3 Self Reflection for Improving a Summary

**Tools used:** Google Colab, Python, Google AI Studio, the Gemini API, and the `google-genai` Python SDK.

**Goal:** Produce an original summary, critique it using explicit requirements, and produce an improved summary.

## Before running

Add `GEMINI_API_KEY` to Colab Secrets using the key icon on the left and enable notebook access. Do not paste the key into the notebook.

```python
!pip -q install -U google-genai

from google import genai
from google.colab import userdata

try:
    api_key = userdata.get("GEMINI_API_KEY")
    if not api_key:
        raise ValueError
except Exception:
    raise ValueError(
        "GEMINI_API_KEY was not found. Add it under the key icon in Colab "
        "and turn on notebook access."
    )

client = genai.Client(api_key=api_key)
MODEL = "gemini-3.5-flash-lite"
print("Gemini client is ready. The API key was loaded from Colab Secrets.")
```

```python
SOURCE = 'The campus library will extend weekday hours from 8 p.m. to 10 p.m. for the two weeks before finals. Weekend hours will stay the same. Students can reserve study rooms online, and the library will post any schedule changes on its website.'

original_prompt = f"""Summarize the source for college students in one short paragraph.

SOURCE:
{SOURCE}"""

original_summary = client.models.generate_content(
    model=MODEL,
    contents=original_prompt,
).text.strip()

print("ORIGINAL PROMPT\n" + original_prompt)
print("\nORIGINAL GEMINI SUMMARY\n" + original_summary)
```

```python
reflection_prompt = f"""You are an editor reviewing a summary for college students.
Compare the original summary only with the source text.

First, write a short critique under these exact headings:
Accuracy
Clarity
Audience
Length
Formatting

Identify unsupported or incorrect claims. Then write an Improved Summary of
35 to 45 words in one paragraph using plain language. It must state the weekday
time change and two-week duration, unchanged weekend hours, online room
reservations, and where updates appear. Do not add benefits or facts that are
not in the source.

SOURCE:
{SOURCE}

ORIGINAL SUMMARY:
{original_summary}"""

review = client.models.generate_content(
    model=MODEL,
    contents=reflection_prompt,
).text.strip()

print("SELF-REFLECTION PROMPT\n" + reflection_prompt)
print("\nCRITIQUE AND IMPROVED SUMMARY\n" + review)
```
